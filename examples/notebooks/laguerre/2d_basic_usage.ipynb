In this example, we will show how to use the `synthetmic` package to generate
a 2D synthetic microstructure using Laguerre diagrams.

Since this is an introduction, we will create a 2D Laguerre diagram
with all grains of equal area.

We first install `synthetmic`.

```python
!pip install synthetmic
```

Next, we import the generator class and functions.

```python
from synthetmic import LaguerreDiagramGenerator
from synthetmic.data.toy import create_data_with_constant_volumes
from synthetmic.plot import plot_2dcells_as_matplotlib_fig
```

Below, we create a unit square with 100 grains. All grains will have area $1/100$ square units (to within 1% tolerance).

Note that you can always create your own custom configuration and are not required to
use the functions provided in `synthetmic`.

```python
config = create_data_with_constant_volumes(
    space_dim=2,
    n_grains=100,
    is_periodic=False,
    random_state=42
)
config
```

```python
config.periodic
```

We then create an instance of the `LaguerreDiagramGenerator` object, and call the `fit` method on the created configuration.

```python
diagram = LaguerreDiagramGenerator(
    tol=1.0,
    n_iter=50,
    damp_param=1.0,
)
diagram.fit(config)
```

Once the diagram is created, we can access many properties, for instance, the areas of the grains, the vertices of the grains, the centroids of the grains, the mesh, etc.

```python
# areas of the grains (all grains have area 1/100 up to the specified tolerance of 1%)
diagram.get_fitted_volumes() 
```

```python
# vertices of the grains
diagram.get_vertices()
```

```python
# centroids of the grains
diagram.get_centroids()
```

```python
diagram.get_mesh()
```

We can plot the generated diagram as follows.

```python
ax = plot_2dcells_as_matplotlib_fig(generator=diagram, colorby=diagram.get_fitted_volumes())
ax.axis("off")
ax.set_aspect("equal")
```

If you want periodicity in any coordinate (say all coordinates),
we just need to modify the configuration and call the `fit` method again.

```python
config.set_periodic((True, True))
diagram.fit(config)
```

We can then plot the diagram accordingly.

```python
ax = plot_2dcells_as_matplotlib_fig(generator=diagram, colorby=diagram.get_fitted_volumes())
ax.axis("off")
ax.set_aspect("equal")
```
