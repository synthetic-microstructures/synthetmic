In this example, we will show how to use the `synthetmic` package to generate
a 3D synthetic microstructure using Voronoi diagrams.

We first install `synthetmic`.

```python
!pip install synthetmic
```

Next, we import the generator class and functions.

```python
from synthetmic import VoronoiDiagramGenerator, DiagramConfig
from synthetmic.data import toy
from synthetmic.plot import plot_cells_as_pyvista_fig
```

Now we create data for the diagram. Below, we create a unit cube with 100 grains.

Note that you can always create your own custom data and are not required to
use the provided functions in `synthetmic`.

```python
domain, _ = toy.create_unit_domain(space_dim=3)
seeds = toy.sample_random_seeds(domain=domain, n_grains=100, random_state=42)

domain, seeds
```

We then create an instance of the `VoronoiDiagramGenerator` object, and call the `fit` method on the created data.

```python
# n_iter=50 is the number of Lloyd iterations. The generated Voronoi diagram is approximately
# a centroidal Voronoi diagram, where each seed is approximately the centroid of its cell.
diagram = VoronoiDiagramGenerator(n_iter=50, damp_param=1.0)
config = DiagramConfig(seeds=seeds, domain=domain, periodic=None)
diagram.fit(config)
```

Once the diagram is created, we can access many properties, for instance, the volumes of the grains, the vertices of the grains, the centroids of the grains, the mesh, etc.

```python
# volumes of the grains
diagram.get_fitted_volumes()
```

```python
# vertices of the grains
diagram.get_vertices()
```

```python
# centroids of the grains
diagram.get_centroids()
```

```python
diagram.get_mesh()
```

We can plot the generated Voronoi diagram as follows.

```python
pl = plot_cells_as_pyvista_fig(generator=diagram, colorby=diagram.get_fitted_volumes(), notebook=True)
pl.show()
```

If you want periodicity in any coordinate (say all coordinates), we can modify the diagram configuration as follows and then call the `fit` method again:

```python
config.set_periodic((True, True, True))
diagram.fit(config)
```

and then plot the corresponding diagram as follows:

```python
pl = plot_cells_as_pyvista_fig(generator=diagram, colorby=diagram.get_fitted_volumes(), notebook=True)
pl.show()
```
